Q1. What is a Convolutional Neural Network (CNN), and how does it differ from traditional fully connected neural networks in terms of architecture and performance on image data?

Ans. A Convolutional Neural Network (CNN) is a specialized type of neural network designed for image data. Unlike traditional fully connected neural networks where every neuron is connected to all neurons in the next layer, CNNs use convolutional layers that apply filters (kernels) to small local regions of an image. This helps the model capture spatial hierarchies such as edges, textures, and shapes. Pooling layers further reduce dimensionality while preserving key features, making the network more efficient. Traditional networks struggle with large image data due to the huge number of parameters, leading to overfitting and computational challenges. CNNs, by contrast, share weights across regions, reducing parameter count and improving performance. As a result, CNNs outperform fully connected networks on image tasks such as classification, object detection, and segmentation, as they mimic how humans perceive visual information.

Q2. Discuss the architecture of LeNet-5 and explain how it laid the foundation for modern deep learning models in computer vision. Include references to its original research paper.

Ans. LeNet-5, proposed by Yann LeCun in 1998, is considered one of the earliest successful CNN architectures. It was primarily designed for handwritten digit recognition (MNIST dataset). The architecture has 7 layers (excluding input) that include convolutional layers, subsampling (pooling) layers, and fully connected layers. Specifically, it consists of two convolutional layers for feature extraction, each followed by average pooling to reduce dimensionality. After feature extraction, the output passes through fully connected layers to perform classification. LeNet-5 introduced key principles still used today: local receptive fields, shared weights, and hierarchical feature extraction. Although small compared to modern networks, it demonstrated that deep learning could outperform traditional methods in computer vision. Its importance lies in inspiring architectures like AlexNet, VGG, and ResNet. The original paper, “Gradient-Based Learning Applied to Document Recognition” (1998), is a landmark in deep learning history.

Q3. Compare and contrast AlexNet and VGGNet in terms of design principles, number of parameters, and performance. Highlight key innovations and limitations of each.

Ans. AlexNet(2012), marked a breakthrough in computer vision by winning the ImageNet competition with significant accuracy improvement. It introduced deeper CNNs, ReLU activation for faster training, and dropout for reducing overfitting. AlexNet had 8 layers (5 convolutional + 3 fully connected) and around 60 million parameters. However, it required powerful GPUs and was prone to overfitting.
VGGNet (2014) from Oxford’s Visual Geometry Group built on AlexNet’s success but focused on simplicity: it used smaller 3×3 convolution filters stacked in depth to improve feature extraction. VGG16 and VGG19 had 16 and 19 layers respectively, with a very large parameter count (138 million for VGG16), making them computationally expensive. AlexNet’s key innovation was ReLU and dropout, while VGGNet’s contribution was proving that deeper and simpler networks improve accuracy. AlexNet was efficient for its time, while VGGNet pushed depth but required more memory and compute power.

Q4. What is transfer learning in the context of image classification? Explain how it helps in reducing computational costs and improving model performance with limited data.

Ans. Transfer learning is a technique where a model trained on a large dataset (e.g., ImageNet with millions of images) is reused for a new but related task. Instead of training a CNN from scratch, we take a pre-trained network and fine-tune it by replacing or retraining its final layers for the new classification problem. This approach is particularly useful when we have limited labeled data, as the pre-trained model already captures general features like edges, shapes, and textures. Only the task-specific features (top layers) need to be adapted. Transfer learning reduces computational costs because it avoids training millions of parameters from scratch and also reduces training time. Additionally, it improves performance since the model leverages knowledge from large-scale datasets. In practice, models like VGG16, ResNet, and Inception are commonly used for transfer learning in medical imaging, plant disease detection, and other fields with limited data.

Q5. Describe the role of residual connections in ResNet architecture. How do they address the vanishing gradient problem in deep CNNs?

Ans. ResNet (Residual Network), introduced in 2015, addressed the major challenge of training very deep networks: the vanishing gradient problem. As networks grow deeper, gradients shrink during backpropagation, making it difficult for earlier layers to learn. ResNet solved this by introducing residual connections, also called “skip connections.” Instead of each layer learning a complete transformation, residual blocks learn only the difference (residual) between the input and desired output. This is done by adding the input directly to the output of a block. This simple addition allows gradients to flow more easily through the network during training, enabling models with over 100 or even 1000 layers to train successfully. The result is improved accuracy and convergence without degradation. ResNet’s innovation made ultra-deep CNNs practical and became the foundation for many modern architectures in computer vision and beyond.



```python
#Q6. Implement the LeNet-5 architecture using TensorFlow or PyTorch to classify the MNIST dataset. Report the accuracy and training time.

import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import time

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = datasets.mnist.load_data()
x_train, x_test = x_train / 255.0, x_test / 255.0

# Reshape for CNN input
x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)

# Build LeNet-5 like model
model = models.Sequential([
    layers.Conv2D(6, (5,5), activation='relu', input_shape=(28,28,1), padding="same"),
    layers.AveragePooling2D(pool_size=(2,2)),
    layers.Conv2D(16, (5,5), activation='relu'),
    layers.AveragePooling2D(pool_size=(2,2)),
    layers.Flatten(),
    layers.Dense(120, activation='relu'),
    layers.Dense(84, activation='relu'),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

start_time = time.time()
history = model.fit(x_train, y_train, epochs=3, validation_split=0.1, verbose=1)
end_time = time.time()

# Evaluate
test_loss, test_acc = model.evaluate(x_test, y_test, verbose=0)

print("Test Accuracy:", test_acc)
print("Training Time (seconds):", round(end_time - start_time, 2))
```

```python
#Q7. Use a pre-trained VGG16 model (via transfer learning) on a small custom dataset (e.g., flowers). Replace the top layers and fine-tune the model.

import tensorflow as tf
from tensorflow.keras.applications import VGG16
from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import pathlib

# Download flower dataset
dataset_url = "https://storage.googleapis.com/download.tensorflow.org/example_images/flower_photos.tgz"
data_dir = tf.keras.utils.get_file('flower_photos', origin=dataset_url, untar=True)
data_dir = pathlib.Path(data_dir)

# Data generator with validation split
train_datagen = ImageDataGenerator(rescale=1./255, validation_split=0.2)

train_gen = train_datagen.flow_from_directory(
    data_dir, target_size=(150,150), batch_size=32, subset='training')

val_gen = train_datagen.flow_from_directory(
    data_dir, target_size=(150,150), batch_size=32, subset='validation')

# Load pre-trained VGG16
base_model = VGG16(weights='imagenet', include_top=False, input_shape=(150,150,3))
base_model.trainable = False  # freeze base layers

# Add new classification head
model = models.Sequential([
    base_model,
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(train_gen.num_classes, activation='softmax')  # automatically matches dataset classes
])

model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Train the model
history = model.fit(train_gen, validation_data=val_gen, epochs=3)

print("Validation Accuracy:", history.history['val_accuracy'][-1])
```

```python
#Q8. Write a program to visualize the filters and feature maps of the first convolutional layer of AlexNet on an example input image.


import torch
# import torchvision.models as models # This line is not needed
import torchvision.transforms as transforms
from PIL import Image
import matplotlib.pyplot as plt
from torchvision.models import alexnet # Import alexnet directly

# Download the image using wget (handles special characters)
!wget -O flower.jpg "https://upload.wikimedia.org/wikipedia/commons/e/e7/Leucanthemum_vulgare_%27Filigran%27_Flower_2200px.jpg"

# Load image
img = Image.open("flower.jpg").convert('RGB')

# Load pre-trained AlexNet
from torchvision.models import AlexNet_Weights
# alexnet = models.alexnet(weights=AlexNet_Weights.IMAGENET1K_V1) # Use the imported alexnet
alexnet_model = alexnet(weights=AlexNet_Weights.IMAGENET1K_V1)
alexnet_model.eval()

# Transform image for AlexNet
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])
img_tensor = transform(img).unsqueeze(0)  # add batch dimension

# Get the first convolutional layer
first_conv_layer = alexnet_model.features[0]

# Visualize the filters
filters = first_conv_layer.weight.data.clone()

fig, axes = plt.subplots(8, 8, figsize=(12,12))  # 64 filters
for i, ax in enumerate(axes.flat):
    if i < filters.shape[0]:
        f = filters[i]
        f_min, f_max = f.min(), f.max()
        f = (f - f_min) / (f_max - f_min)  # normalize
        ax.imshow(f.permute(1, 2, 0))
        ax.axis('off')
plt.suptitle("First Conv Layer Filters of AlexNet")
plt.show()

# Get feature maps by forwarding the image through the first conv layer
with torch.no_grad():
    feature_maps = first_conv_layer(img_tensor)

# Visualize feature maps
fig, axes = plt.subplots(8, 8, figsize=(12,12))
for i, ax in enumerate(axes.flat):
    if i < feature_maps.shape[1]:
        fmap = feature_maps[0, i, :, :].cpu()
        ax.imshow(fmap, cmap='viridis')
        ax.axis('off')
plt.suptitle("Feature Maps of First Conv Layer of AlexNet")
plt.show()
```

```python
#Q9. Train a GoogLeNet (Inception v1) or its variant using a standard dataset like CIFAR-10. Plot the training and validation accuracy over epochs and analyze overfitting or underfitting.


import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt

# Device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print("Using device:", device)

# CIFAR-10 dataset (128x128 input for faster training)
transform_train = transforms.Compose([
    transforms.Resize((128,128)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize((0.4914,0.4822,0.4465),(0.2023,0.1994,0.2010))
])

transform_test = transforms.Compose([
    transforms.Resize((128,128)),
    transforms.ToTensor(),
    transforms.Normalize((0.4914,0.4822,0.4465),(0.2023,0.1994,0.2010))
])

train_dataset = torchvision.datasets.CIFAR10(root='./data', train=True, download=True, transform=transform_train)
train_loader = torch.utils.data.DataLoader(train_dataset, batch_size=32, shuffle=True)

test_dataset = torchvision.datasets.CIFAR10(root='./data', train=False, download=True, transform=transform_test)
test_loader = torch.utils.data.DataLoader(test_dataset, batch_size=32, shuffle=False)

# Load pretrained GoogLeNet with auxiliary heads
from torchvision.models import googlenet, GoogLeNet_Weights
model = googlenet(weights=GoogLeNet_Weights.IMAGENET1K_V1, aux_logits=True)
model.to(device)

# Loss and optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# Training settings
num_epochs = 3
train_acc_list = []
val_acc_list = []

for epoch in range(num_epochs):
    model.train()
    correct = 0
    total = 0
    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(images)  # may be GoogLeNetOutputs object

        # Compute main + auxiliary loss
        if hasattr(outputs, 'logits'):  # training returns GoogLeNetOutputs
            loss = criterion(outputs.logits, labels)
            if hasattr(outputs, 'aux1') and outputs.aux1 is not None:
                loss += 0.3*criterion(outputs.aux1, labels)
            if hasattr(outputs, 'aux2') and outputs.aux2 is not None:
                loss += 0.3*criterion(outputs.aux2, labels)
            preds = outputs.logits.argmax(dim=1)
        else:  # inference returns Tensor
            loss = criterion(outputs, labels)
            preds = outputs.argmax(dim=1)

        loss.backward()
        optimizer.step()

        total += labels.size(0)
        correct += preds.eq(labels).sum().item()

    train_acc = 100 * correct / total
    train_acc_list.append(train_acc)

    # Validation
    model.eval()
    correct = 0
    total = 0
    with torch.no_grad():
        for images, labels in test_loader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            # handle Tensor or GoogLeNetOutputs
            if hasattr(outputs, 'logits'):
                preds = outputs.logits.argmax(dim=1)
            else:
                preds = outputs.argmax(dim=1)
            total += labels.size(0)
            correct += preds.eq(labels).sum().item()
    val_acc = 100 * correct / total
    val_acc_list.append(val_acc)

    print(f"Epoch [{epoch+1}/{num_epochs}], Train Acc: {train_acc:.2f}%, Val Acc: {val_acc:.2f}%")

# Plot accuracy
plt.figure(figsize=(8,5))
plt.plot(range(1,num_epochs+1), train_acc_list, label='Train Accuracy')
plt.plot(range(1,num_epochs+1), val_acc_list, label='Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy (%)')
plt.title('GoogLeNet Training and Validation Accuracy on CIFAR-10')
plt.legend()
plt.show()




```

Q10. You are working in a healthcare AI startup. Your team is tasked with developing a system that automatically classifies medical X-ray images into normal, pneumonia, and COVID-19. Due to limited labeled data, what approach would you suggest using among CNN architectures discussed (e.g., transfer learning with ResNet or Inception variants)? Justify your approach and outline a deployment strategy for production use.

Ans. In this problem, we have to classify medical X-ray images into normal, pneumonia, and COVID-19. But the challenge is that there is limited labeled data. If we try to train a CNN from scratch, it will not work well because CNNs need a large amount of data and high computing power. So, the best approach here is to use transfer learning.

Among the models we studied, I will choose ResNet. ResNet is very powerful because of its residual (skip) connections, which help the network avoid the vanishing gradient problem. This makes it easier to train even very deep models. Also, ResNet models are already trained on ImageNet with millions of images, so they know basic image features like edges, textures, and shapes. We can reuse this knowledge and only change the last layer for 3 classes: normal, pneumonia, and COVID-19. This saves a lot of time, gives better accuracy, and reduces the cost of training.

For deployment, I would use this step-by-step plan:

1. Data preparation – Collect X-ray images, resize, normalize, and use augmentation (rotate, flip) to create more samples.

2. Model training – Take pre-trained ResNet, freeze base layers, replace final layer with 3 outputs, fine-tune on the medical dataset.

3. Validation – Test accuracy with hospital-provided validation data to ensure reliability.

4. Deployment – Host the trained model on a cloud server or in the hospital’s system. Doctors can upload X-ray images and get predictions instantly.

5. Monitoring & updates – Keep collecting new data, retrain, and improve accuracy over time.
